```python
# Prepare text for N-gram analysis
def preprocess_text_for_ngrams(text):
    text = text.lower()
    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    words = word_tokenize(text)
    # Remove stopwords and single character words
    filtered_words = [word for word in words if word not in stopwords.words('english') and len(word) > 1]
    return filtered_words

# Combine all messages into a single string (excluding group notifications)
all_messages_combined = " ".join(df[df['user'] != 'group_notification']['message'].astype(str))
preprocessed_words = preprocess_text_for_ngrams(all_messages_combined)

print(f"Total preprocessed words for N-gram analysis: {len(preprocessed_words)}")
```

```python
# Generate and count Bigrams (2-word phrases)
bigrams = list(ngrams(preprocessed_words, 2))
bigram_counts = Counter(bigrams)

# Get top 20 bigrams
top_20_bigrams = bigram_counts.most_common(20)

print("\nTop 20 Most Common Bigrams:")
for bigram, count in top_20_bigrams:
    print(f"{bigram}: {count}")

# Visualize Top 20 Bigrams
bigrams_df = pd.DataFrame(top_20_bigrams, columns=['Bigram', 'Count'])
bigrams_df['Bigram_Str'] = bigrams_df['Bigram'].apply(lambda x: ' '.join(x))

plt.figure(figsize=(12, 8))
sns.barplot(x='Count', y='Bigram_Str', data=bigrams_df, hue='Bigram_Str', palette='viridis', legend=False)
plt.title('Top 20 Most Common Bigrams')
plt.xlabel('Count')
plt.ylabel('Bigram')
plt.tight_layout()
plt.show()
```

```python
# Generate and count Trigrams (3-word phrases)
trigrams = list(ngrams(preprocessed_words, 3))
trigram_counts = Counter(trigrams)

# Get top 20 trigrams
top_20_trigrams = trigram_counts.most_common(20)

print("\nTop 20 Most Common Trigrams:")
for trigram, count in top_20_trigrams:
    print(f"{trigram}: {count}")

# Visualize Top 20 Trigrams
trigrams_df = pd.DataFrame(top_20_trigrams, columns=['Trigram', 'Count'])
trigrams_df['Trigram_Str'] = trigrams_df['Trigram'].apply(lambda x: ' '.join(x))

plt.figure(figsize=(12, 8))
sns.barplot(x='Count', y='Trigram_Str', data=trigrams_df, hue='Trigram_Str', palette='magma', legend=False)
plt.title('Top 20 Most Common Trigrams')
plt.xlabel('Count')
plt.ylabel('Trigram')
plt.tight_layout()
plt.show()
```

### User-to-User Interaction Analysis

```python
# Get a list of all unique users (excluding 'group_notification')
all_users = df['user'].unique().tolist()
all_users = [user for user in all_users if user != 'group_notification']

# Create a regex pattern to find mentions of any user in a message
# The pattern looks for the exact user name followed by a non-alphanumeric character or end of string
mention_pattern = r'\b(' + '|'.join(re.escape(user) for user in all_users) + r')\b'

def find_mentions(message, sender_user):
    mentions = []
    if isinstance(message, str):
        found_mentions = re.findall(mention_pattern, message, re.IGNORECASE)
        for mentioned_user in found_mentions:
            # Try to match the exact case from all_users list
            exact_match = next((u for u in all_users if u.lower() == mentioned_user.lower()), None)
            if exact_match and exact_match != sender_user: # Exclude self-mentions
                mentions.append(exact_match)
    return list(set(mentions)) # Return unique mentions

# Apply the function to create a new 'mentions' column
df['mentions'] = df.apply(lambda row: find_mentions(row['message'], row['user']), axis=1)

# Filter messages that actually contain mentions
mention_interactions = df[df['mentions'].apply(len) > 0]

print(f"Total messages with mentions: {mention_interactions.shape[0]}")

if not mention_interactions.empty:
    print("\nExamples of messages with mentions:")
    display(mention_interactions[['user', 'message', 'mentions']].head())
else:
    print("No explicit user mentions found in the messages.")
```

Now, let's visualize the most frequent mention interactions.

```python
# Analyze mention frequency
mention_pairs = []
for _, row in mention_interactions.iterrows():
    sender = row['user']
    for receiver in row['mentions']:
        mention_pairs.append(f"{sender} -> {receiver}")

mention_counts = Counter(mention_pairs)

if mention_counts:
    top_mention_interactions = mention_counts.most_common(10)
    print("\nTop 10 Mention Interactions:")
    for interaction, count in top_mention_interactions:
        print(f"{interaction}: {count}")

    # Prepare data for plotting
    mentions_df = pd.DataFrame(top_mention_interactions, columns=['Interaction', 'Count'])

    plt.figure(figsize=(10, 6))
    sns.barplot(x='Count', y='Interaction', data=mentions_df, hue='Interaction', palette='coolwarm', legend=False)
    plt.title('Top 10 User Mention Interactions')
    plt.xlabel('Number of Mentions')
    plt.ylabel('Interaction (Sender -> Receiver)')
    plt.tight_layout()
    plt.show()
else:
    print("No sufficient mention interactions to visualize.")
```

### Sentiment Analysis

```python
!pip install nltk --quiet
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import matplotlib.pyplot as plt
import seaborn as sns

# Download the VADER lexicon for sentiment analysis
print("Attempting to download NLTK 'vader_lexicon'...")
nltk.download('vader_lexicon', quiet=False)
print("NLTK 'vader_lexicon' download attempt complete.")

print("NLTK setup for sentiment analysis completed.")
```

```python
# Initialize the VADER sentiment intensity analyzer
sia = SentimentIntensityAnalyzer()

# Function to get sentiment scores
def get_sentiment_scores(text):
    if isinstance(text, str):
        return sia.polarity_scores(text)
    return {'neg': 0, 'neu': 0, 'pos': 0, 'compound': 0}

# Apply sentiment analysis to each message
df['sentiment_scores'] = df['message'].apply(get_sentiment_scores)

# Extract compound score for overall sentiment
df['compound_sentiment'] = df['sentiment_scores'].apply(lambda x: x['compound'])

# Categorize sentiment based on compound score
def categorize_sentiment(score):
    if score >= 0.05:
        return 'Positive'
    elif score <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'

df['sentiment_category'] = df['compound_sentiment'].apply(categorize_sentiment)

print("Sentiment analysis completed. Displaying sentiment categories distribution:")
display(df['sentiment_category'].value_counts())
display(df[['user', 'message', 'sentiment_category', 'compound_sentiment']].head())
```

```python
# Visualize the distribution of sentiment categories
sentiment_counts = df['sentiment_category'].value_counts().reset_index()
sentiment_counts.columns = ['Sentiment', 'Count']

plt.figure(figsize=(8, 6))
sns.barplot(x='Sentiment', y='Count', data=sentiment_counts, hue='Sentiment', palette='coolwarm', legend=False)
plt.title('Distribution of Sentiment in Chat Messages')
plt.xlabel('Sentiment Category')
plt.ylabel('Number of Messages')
plt.tight_layout()
plt.show()
```
